```python
import pandas as pd
import numpy as np
import os
```

```python
# Nama file dataset
file_path = "klasifikasimhs.csv"

# Memeriksa apakah file tersedia
if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    print("Dataset berhasil dibaca!")
else:
    print("File tidak ditemukan. Periksa lokasi dataset.")
```

```python
# Informasi dataset
nama_dataset = "Klasifikasi Mahasiswa"
sumber_dataset = "Isi dengan sumber asli dataset"
lisensi_dataset = "Belum diketahui"

print("Nama Dataset :", nama_dataset)
print("Sumber       :", sumber_dataset)
print("Lisensi      :", lisensi_dataset)
```

```python
# Menampilkan 5 baris pertama
display(df.head())

display(df.tail())
```

```python
jumlah_baris, jumlah_kolom = df.shape

print("Jumlah baris  :", jumlah_baris)
print("Jumlah kolom  :", jumlah_kolom)
print("Total data    :", df.size)
```

```python
print("Nama Kolom:")
print(df.columns.tolist())

print("\nTipe Data Setiap Kolom:")
print(df.dtypes)
df.info()
```

```python
# Menghitung jumlah data kosong setiap kolom
data_kosong = df.isnull().sum()

print("Jumlah Data Kosong per Kolom:")
print(data_kosong)

print("\nTotal Data Kosong:", df.isnull().sum().sum())
```

```python
# Statistik deskriptif untuk semua kolom
display(df.describe(include="all").T)
```

```python
jumlah_duplikat = df.duplicated().sum()

print("Jumlah baris duplikat:", jumlah_duplikat)
```

```python
for kolom in df.columns:
    print(f"\nKolom: {kolom}")
    print("Jumlah nilai unik:", df[kolom].nunique())
    print("Contoh nilai:", df[kolom].unique()[:10])
```

```python
ringkasan = pd.DataFrame({
    "Informasi": [
        "Nama Dataset",
        "Jumlah Baris",
        "Jumlah Kolom",
        "Jumlah Data Kosong",
        "Jumlah Baris Duplikat",
        "Jumlah Kolom Numerik",
        "Jumlah Kolom Kategorikal"
    ],
    "Hasil": [
        nama_dataset,
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df.select_dtypes(include="number").shape[1],
        df.select_dtypes(include="object").shape[1]
    ]
})

display(ringkasan)
```
